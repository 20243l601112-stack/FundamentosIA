# Problema: Realizar un sistema que identifique si una planta necesita riego para activar el sensor.


```python
print("Sensor de riego")
nombre=input("Escriba el nombre de la planta")
humedad=float(input("Escribe el nivel de humedad de la planta"))
```

```python
print ("Datos almacenados")
print ("Nombre de la planta:", nombre)
print ("Nivel de humedad:", humedad, "%")
```

```python
if humedad<40:
    print("La planta necesita riego")
elif humedad>=40:
    print("La planta no necesita riego")
```
