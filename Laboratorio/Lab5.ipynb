```python
# Situacion: un agente inteligente se encuentra en A y necesita llegar a B
```

```python
#Libreria utilizada para graficar
import matplotlib.pyplot as plt

import networkx as nx
```

```python
#creamos una instancia de la clase  graph
#Graph  es una coleccion de nodos
grafo = nx.Graph()
```

```python
#Creamos el grafo utilizando el metodo add_edges
grafo.add_edges_from(
    [
        ("A","B"),
        ("A","C"),
        ("A","E"),
        ("B","D"),
        ("D","F"),
        ("E","G"),
        ("E","F")
    ]
)
```

```python
#Mostrar el grafo

nx.draw(
    grafo,
    with_labels = True,
    node_size=1000
)
plt.show
```

```python
# Mostrar las rutas posibles del grafo
rutas= list (
    nx.all_simple_paths(grafo, source="A", target="G")
)
```

```python
#imprimir las rutas
print("Rutas encontradas:")
for i, ruta in enumerate(rutas, start=1):
    print(f"Ruta {i}: {ruta}")
```

```python
#Mostrar el total de pasos de cada ruta
pasos1 = len (rutas[0])-1
print("Pasos de la ruta 1: ", pasos1)
pasos2 = len(rutas[1])-1
print("Pasos de la ruta 2: ", pasos2)
```

```python
# Aplicamos la logica proposicional
P = pasos1 < pasos2
Q = pasos2 < pasos1

if P==True:
    print("P = True")
    print("La ruta 1 es mejor opcion.")
elif Q==True:
    print("Q = True")
    print("La ruta 2 es mejor opcion.")
else:
    print("P = False")
    print("Q = False")
    print("Las dos rutas son iguales de buenas, haz lo que quieras!")
```
